```python
#Question 3 - For missing values find and replace with arithmetic mean of the attributes
df.isnull().any()
```

```python
#Question 3 - Replace missing values with mean 

newdf = df.replace(-1,df[["protein","fat", "sodium", "fiber","sugars","potass","vitamins"]].mean())
```

```python
#Question 4 - Find and replace outliers with median 

for col in ["protein","fat", "sodium", "fiber","carbo","sugars","potass","vitamins"]:

    Q1 = df[col].quantile(0.25)


    Q3 = df[col].quantile(0.75)


    IQR = Q3 - Q1


    upper = Q3+ 1.5 * IQR
    lower = Q1- 1.5*IQR
    df.loc[df[col]>upper, col] = np.NaN
    df.loc[df[col]<lower, col] = np.NaN
    df[col].fillna(df[col].median(), inplace=True)
```
