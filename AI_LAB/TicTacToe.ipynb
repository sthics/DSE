```python
from easyAI import TwoPlayerGame, Human_Player, AI_Player, Negamax, solve_with_depth_first_search , solve_with_iterative_deepening
```

```python
class TicTacToe( TwoPlayerGame ):
    """ The board positions are numbered as follows:
            7 8 9
            4 5 6
            1 2 3
    """    

    def __init__(self, players):
        """
         Initialization of the game
        """
        self.players = players
        self.board = [0 for i in range(9)]
        self.current_player = 1 # player 1 starts.
    
    def possible_moves(self):
        """
        Returns of all moves allowed excluding already played moves
        """
        return [i+1 for i,e in enumerate(self.board) if e==0]
    
    def make_move(self, move):
        """
        Transforms the game according to the move
        """
        self.board[int(move)-1] = self.current_player
    
    def lose(self):
        """ Has the opponent "three in line ?" """
        return any( [all([(self.board[c-1]== self.opponent_index)
                      for c in line])
                      for line in [[1,2,3],[4,5,6],[7,8,9], 
                                   [1,4,7],[2,5,8],[3,6,9], 
                                   [1,5,9],[3,5,7]]]) 
        
    def is_over(self):
        """
        Check whether the game has ended
        """
        return (self.possible_moves() == []) or self.lose()
        
    def show(self):
        """
        Displays the board after each move
        """
        print ('\n'+'\n'.join([
                        ' '.join([['.','O','X'][self.board[3*j+i]]
                        for i in range(3)])
                 for j in range(3)]) )
                 
    def scoring(self):
        """
        Handles the scoring
        """
        return -100 if self.lose() else 0
```

```python
ai = Negamax(6)
#A Tic Tac Toe game between AI and a human player
TicTacToe( [Human_Player(),AI_Player(ai)]).play()
```

```python
#Exploring using iterative deepining using a ai against a human playing the best possible game
#d:depth a:alpha m:best moves played?
result, depth, move=solve_with_iterative_deepening(TicTacToe( [Human_Player(),AI_Player(ai)]),ai_depths=range(2,12),verbose=True,win_score=90)
```

```python
print(f"Result for first payer: {result}\nDepth: {depth}\nBest move for first player: {move}")
#It is impossible for anyone to win in a perfect match of tic tac toe
#thus it always results in a draw
```
